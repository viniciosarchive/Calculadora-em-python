```python
num1= float (input("insira um número"))
num2= float(input("insira outro numero"))
operacao= input("qual a operação? soma, subtracao, multiplicacao, divisao ou resto")
soma= num1 + num2
subtracao= num1 - num2
divisao= num1 / num2
multiplicacao= num1 * num2
resto= num1 % num2
if operacao == "soma":
  print("O resultado é:", (soma))

elif operacao == "subtracao":
  print("O resultado é:", (subtracao))

elif operacao == "divisao":
  print("O resultado é:", (divisao))

elif operacao == "multiplicacao":
  print("O resultado é:", (multiplicacao))

elif operacao == "resto":
  print("O resultado é:", (resto))

else:
  print("Operação Inválida")


```
